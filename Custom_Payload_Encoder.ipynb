This project demonstrates practical payload encoding and string obfuscation techniques used for studying static signature evasion in controlled lab environments.

# 1. Import Required Libraries

```python
import base64
import codecs
import random
import string

from datetime import datetime
```

# 2. Payload Encoding Modules

This section implements multiple payload encoding techniques:
- Base64
- XOR
- ROT13

```python
def base64_encode(data):
    return base64.b64encode(data.encode()).decode()

def base64_decode(data):
    return base64.b64decode(data.encode()).decode()
```

```python
def xor_encode(data, key):
    return ''.join(chr(ord(c) ^ key) for c in data)

def xor_decode(data, key):
    return ''.join(chr(ord(c) ^ key) for c in data)
```

```python
def rot13_encode(data):
    return codecs.encode(data, 'rot_13')
```

# 3. String Obfuscation Techniques

This section applies multiple string obfuscation methods to transform payloads and reduce signature visibility.

```python
def random_insertion(data):

    result = ""

    for ch in data:
        result += ch
        result += random.choice(string.ascii_letters)

    return result
```

```python
def split_concat(data):

    parts = [data[i:i+3] for i in range(0, len(data), 3)]

    return " + ".join(f'"{p}"' for p in parts)
```

```python
def escape_obfuscation(data):
    return ''.join(f'\\x{ord(c):02x}' for c in data)
```

# 4. Detection Simulation Engine

This module simulates basic static signature detection using keyword matching.

```python
SIGNATURES = [
    "Invoke",
    "Payload",
    "TestFunction"
]

def detect_signature(data):

    detected = []

    for sig in SIGNATURES:

        if sig.lower() in data.lower():
            detected.append(sig)

    return detected
```

# 5. Main Execution Logic

This section processes payloads through encoding and obfuscation pipelines.

```python
payload = "Invoke-TestFunction"

print("=" * 50)
print("CUSTOM PAYLOAD ENCODER")
print("=" * 50)

print("\nOriginal Payload:")
print(payload)
```

# 6. Encoding & Obfuscation Tests

```python
base64_payload = base64_encode(payload)

xor_payload = xor_encode(payload, 5)

rot13_payload = rot13_encode(payload)

random_payload = random_insertion(payload)

split_payload = split_concat(payload)

escape_payload = escape_obfuscation(payload)
```

# 7. Generated Payload Outputs

```python
print("\n[Base64]")
print(base64_payload)

print("\n[XOR]")
print(xor_payload)

print("\n[ROT13]")
print(rot13_payload)

print("\n[Random Insertion]")
print(random_payload)

print("\n[Split Concatenation]")
print(split_payload)

print("\n[Escape Obfuscation]")
print(escape_payload)
```

# 8. Evasion Testing Results

This section compares signature detection between original and transformed payloads.

```python
tests = {
    "Base64": base64_payload,
    "XOR": xor_payload,
    "ROT13": rot13_payload,
    "Random": random_payload,
    "Split": split_payload,
    "Escape": escape_payload
}

print("=" * 50)
print("DETECTION RESULTS")
print("=" * 50)

for name, value in tests.items():

    result = detect_signature(value)

    if result:
        print(f"{name}: DETECTED -> {result}")
    else:
        print(f"{name}: BYPASSED")
```

# 9. Multi-Layer Encoding

This section applies multiple encoding techniques sequentially to simulate advanced payload transformations.

```python
def multi_layer_encode(payload, xor_key):

    # Layer 1 → ROT13
    step1 = rot13_encode(payload)

    # Layer 2 → XOR
    step2 = xor_encode(step1, xor_key)

    # Layer 3 → Base64
    final_payload = base64_encode(step2)

    return {
        "ROT13": step1,
        "XOR": step2,
        "BASE64_FINAL": final_payload
    }
```

```python
multi_payload = multi_layer_encode(payload, 7)

print("=" * 50)
print("MULTI-LAYER ENCODING")
print("=" * 50)

for layer, value in multi_payload.items():

    print(f"\n[{layer}]")
    print(value)
```

```python
final_result = detect_signature(
    multi_payload["BASE64_FINAL"]
)

print("\nDetection Result:")

if final_result:
    print("DETECTED:", final_result)
else:
    print("BYPASSED")
```

# 10. Detection Scoring System

This section evaluates the effectiveness of each encoding and obfuscation technique against simulated signature detection.

```python
def calculate_detection_score(data):

    signatures_found = detect_signature(data)

    total_signatures = len(SIGNATURES)

    matched = len(signatures_found)

    score = int((matched / total_signatures) * 100)

    return {
        "matched_signatures": signatures_found,
        "score": score
    }
```

```python
all_payloads = {
    "Original": payload,
    "Base64": base64_payload,
    "XOR": xor_payload,
    "ROT13": rot13_payload,
    "RandomInsertion": random_payload,
    "SplitConcat": split_payload,
    "EscapeObfuscation": escape_payload,
    "MultiLayer": multi_payload["BASE64_FINAL"]
}
```

```python
print("=" * 80)
print("\t\t\t DETECTION EFFECTIVENESS ANALYSIS ")
print("=" * 80)

for name, data in all_payloads.items():

    result = calculate_detection_score(data)

    score = result["score"]

    if score > 0:
        status = "DETECTED"
    else:
        status = "BYPASSED"

    print(f"\nTechnique: {name}")
    print(f"Status: {status}")
    print(f"Detection Score: {score}%")
    print(f"Matched Signatures: {result['matched_signatures']}")
```

# 11. Comparative Analysis Table

```python
from tabulate import tabulate

table_data = []

for name, data in all_payloads.items():

    result = calculate_detection_score(data)

    score = result["score"]

    if score > 0:
        status = "DETECTED"
    else:
        status = "BYPASSED"

    table_data.append([
        name,
        status,
        f"{score}%"
    ])

print(tabulate(
    table_data,
    headers=["Technique", "Status", "Detection Score"],
    tablefmt="grid"
))
```
